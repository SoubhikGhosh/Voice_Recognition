```python
import os
import wave
import json
from vosk import Model, KaldiRecognizer
import subprocess

# Path to your model (download from Vosk website if you haven't already)
model_path = "../Models/vosk-model-en-in-0.5"  # Update with your model path

# Function to convert m4a to wav
def convert_m4a_to_wav(input_file, output_file):
    command = f"ffmpeg -i {input_file} {output_file}"
    subprocess.run(command, shell=True)

# Set up your audio file and output
audio_file = "../Data/Ravi.m4a"  # Input m4a file
wav_file = "Ravi.wav"    # Output wav file (converted from m4a)

# Convert audio from m4a to wav (if not already in wav format)
convert_m4a_to_wav(audio_file, wav_file)

# Check if the wav file exists
if not os.path.exists(wav_file):
    print(f"Error: {wav_file} not found!")
    exit()

# Load the Vosk model
model = Model(model_path)

# Open the WAV file for reading
wf = wave.open(wav_file, "rb")

# Check the sample rate
sample_rate = wf.getframerate()
print(f"Sample Rate: {sample_rate}")

# Create a recognizer instance
recognizer = KaldiRecognizer(model, sample_rate)

# Transcribe the audio
text = ""
while True:
    data = wf.readframes(4000)  # Read larger frames for better processing
    if len(data) == 0:
        break
    if recognizer.AcceptWaveform(data):
        result = recognizer.Result()
        result_json = json.loads(result)
        print(f"Intermediate result: {result_json.get('text', '')}")  # Print intermediate results
        text += result_json.get("text", "") + " "

# Print the final transcribed text
print(f"Final Transcription: {text}")

# Clean up the temporary WAV file
os.remove(wav_file)
```

```python
import subprocess
import os

# Define paths
audio_file = "../Data/Keshab.m4a"
output_file = "KOut.txt"
model_paths = ["../Models/vosk-model-en-in-0.5/", "../Models/vosk-model-small-en-in-0.4/"]

for model_path in model_paths:
    # Run vosk-transcriber using subprocess
    command = f"vosk-transcriber -i {audio_file} -o {output_file} -m {model_path}"
    print(command)
    subprocess.run(command, shell=True, check=True)

    # Read the output transcription from the text file
    if os.path.exists(output_file):
        with open(output_file, 'r') as file:
            transcription = file.read()
            print("Transcription:")
            print(transcription)

        # Delete the output text file after reading
        os.remove(output_file)
    else:
        print(f"Error: {output_file} not found!")
```
