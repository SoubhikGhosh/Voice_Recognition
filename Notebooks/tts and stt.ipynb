## TTS

```python
import AVFoundation

def text_to_speech(text):
    # Initialize the speech synthesizer
    synthesizer = AVFoundation.AVSpeechSynthesizer.alloc().init()

    # Choose voice with Indian English accent
    voice = AVFoundation.AVSpeechSynthesisVoice.voiceWithLanguage_("en-IN")

    # Prepare the speech utterance with the text and set the voice
    utterance = AVFoundation.AVSpeechUtterance.alloc().initWithString_(text)
    utterance.setValue_forKey_(voice, "voice")  # Correct way to set the voice
    utterance.setValue_forKey_(0.5, "rate")  # Adjust speech rate

    # Speak the utterance
    synthesizer.speakUtterance_(utterance)

# Example usage
text_to_speech("Hello! This is a sample text being spoken in Indian English.")
```

```python
import pyttsx3

def text_to_speech(text):
    engine = pyttsx3.init()
    engine.setProperty("voice", "com.apple.voice.compact.mr-IN.Ananya")  # Use Indian English voice
    engine.setProperty("rate", 170)  # Adjust speech rate
    engine.say(text)
    engine.runAndWait()

text_to_speech("নবেন্দুশেখরের সহিত অরুণলেখার যখন বিবাহ হইল, তখন হোমধূমের অন্তরাল হইতে ভগবান প্রজাপতি ঈষৎ একটু হাস্য করিলেন। হায়, প্রজাপতির পক্ষে যাহা খেলা আমাদের পক্ষে তাহা সকল সময়ে কৌতুকের নহে।")
```

## STT

```python
import Speech
import AVFoundation
import time

def speech_to_text():
    # Initialize the speech recognizer
    recognizer = Speech.SFSpeechRecognizer.alloc().initWithLocale_(AVFoundation.NSLocale.localeWithLocaleIdentifier_("en-IN"))

    if not recognizer.isAvailable():
        print("Speech recognition is not available.")
        return

    # Create a request for speech recognition
    recognition_request = Speech.SFSpeechAudioBufferRecognitionRequest.alloc().init()

    # Use the device's default audio input (microphone)
    audio_engine = AVFoundation.AVAudioEngine.alloc().init()
    input_node = audio_engine.inputNode()

    # Configure the audio format
    recording_format = input_node.outputFormatForBus_(0)
    input_node.installTapOnBus_bufferSize_format_block_(
        0, 1024, recording_format, lambda buffer, time: recognition_request.appendAudioPCMBuffer_(buffer)
    )

    # Start the audio engine to begin recording
    audio_engine.prepare()
    audio_engine.startAndReturnError_(None)

    # Start recognition task
    recognition_task = recognizer.recognitionTaskWithRequest_resultHandler_(recognition_request, lambda result, error: print(result.bestTranscription().formattedString()) if result else print("Error:", error))

    # Allow time for speech input
    print("Please start speaking...")
    time.sleep(10)

    # Stop recording
    audio_engine.stop()
    recognition_request.endAudio()

# Example usage
speech_to_text()
```

```python
import speech_recognition as sr
from IPython.display import display
from ipywidgets import Output
import os

# Output widget to display real-time transcriptions
output_widget = Output()

def recognize_speech_from_mic():
    """Recognizes speech in real-time from the microphone."""
    recognizer = sr.Recognizer()
    mic = sr.Microphone()

    with mic as source:
        recognizer.adjust_for_ambient_noise(source)
        print("Listening... Speak now!")

        try:
            while True:
                audio = recognizer.listen(source, timeout=5)
                text = recognizer.recognize_google(audio, language="en-IN")  # Indian English
                with output_widget:
                    print(f"You said: {text}")
                    display(output_widget)  # Display in Jupyter output widget
        except sr.WaitTimeoutError:
            print("Timeout: No speech detected.")
        except sr.UnknownValueError:
            print("Could not understand audio.")
        except sr.RequestError:
            print("Could not connect to Google Speech API.")

def recognize_speech_from_wav(file_path):
    """Recognizes speech from a .wav file."""
    recognizer = sr.Recognizer()
    
    # Check if the file exists
    if not os.path.exists(file_path):
        print("File not found:", file_path)
        return
    
    with sr.AudioFile(file_path) as source:
        audio = recognizer.record(source)  # Read entire audio file

    try:
        text = recognizer.recognize_google(audio, language="en-IN")
        print("Transcription from file:", text)
    except sr.UnknownValueError:
        print("Could not understand the audio in file.")
    except sr.RequestError:
        print("Could not connect to Google Speech API.")

# Example usage:
# recognize_speech_from_mic()  # Uncomment to recognize speech from microphone
recognize_speech_from_wav("../Data/RaviTest.wav")  # Uncomment to transcribe a WAV file
```
